```python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
original = np.load(
    "pulse_before_inference_log/0.npy"
).squeeze(1)
```

```python
for i in range(1,10):

    sucess = np.load(f"pulse_before_inference_log/{i}.npy").squeeze(
        1
    )

    max_original = []
    max_sucess = []

    for a,b in zip(original,sucess):
        max_original.append(a.argmax())
        max_sucess.append(b.argmax())
        
    df = pd.DataFrame({
        "original_classes":max_original,
        "attack_classes":max_sucess,
    })

    pd.crosstab(df.original_classes,df.attack_classes).to_csv(f"cross_{i}.csv")
    

    plt.figure(figsize=(10, 10))
    plt.title(i)
    cross = pd.crosstab(df.original_classes, df.attack_classes)
    sns.heatmap(cross, fmt='d', cmap='Blues')
    plt.tight_layout()
    plt.savefig(f"{i}.png")
    # plt.show()
```
