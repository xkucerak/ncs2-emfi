```python
import torch
```

```python
from torchvision.models import resnet18, ResNet18_Weights

model = resnet18(weights=ResNet18_Weights.IMAGENET1K_V1).eval()

torch.onnx.export(
    model,
    torch.ones(1, 3, 224, 224),
    "resnet18.onnx",
    dynamo=False,
)
```
