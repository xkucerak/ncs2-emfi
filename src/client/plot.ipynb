```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
folder = Path("../../results/exploratory/").resolve()

for i in folder.iterdir():
    
    results_path = (i / "results.csv")
    settings_path = (i / "settings.csv")
    
    if results_path.exists():
        for type in ["top_1", "top_5"]:
            df = pd.read_csv(results_path).replace(-1, np.nan)
            plt.figure(figsize=(10, 4))
            print()
            plt.title(
                type.upper().replace("_", " ")
                + " | "
                + pd.read_csv(settings_path)["cfg_model"].item()
                + " | Samples: "
                + str(len(df[type]))
                + " | Failed: "
                + str(df[type].isna().sum().item())
            )
            sns.histplot(df[type], bins=2**6)
            plt.tight_layout()
            plt.savefig((i/f"{type}.svg"))
```

```python
for i in range(2, 5):
    df = pd.read_csv(
        f"../../results/exploratory/{str(i).zfill(4)}_spot/results.csv"
    ).replace(-1, np.nan)

    print(
        pd.read_csv(f"../../results/exploratory/{str(i).zfill(4)}_spot/settings.csv")[
            "cfg_model"
        ].item()
    )

    print(
        df["top_1"]
        .value_counts(dropna=False)
        .sort_index(ascending=False, na_position="first")
        .to_markdown()
    )

    print("*" * 10)
    # print(df["top_1"].isna().sum())
```

```python
for i in range(33, 34):
    for type in ["top_1", "top_5"]:
        results = pd.read_csv(
            f"../../results/exploratory/{str(i).zfill(4)}_spot/results.csv"
        ).replace(-1, np.nan)
        settings = pd.read_csv(
            f"../../results/exploratory/{str(i).zfill(4)}_spot/settings.csv"
        )

        target = settings["cfg_" + type].item()

        results[type + "_diff"] = (df[type] - target) / target
        sns.histplot(results[type + "_diff"], bins=100)

        print(pd.cut(results[type + "_diff"], [0, 0.01, 0.1, 1, float("inf")]))
```
