```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
for i in range(2,5):

    df = pd.read_csv(
        f"../../results/exploratory/{str(i).zfill(4)}_spot/results.csv"
    ).replace(-1, np.nan)
    plt.figure(figsize=(10, 3))
    print()
    plt.title(
        pd.read_csv(f"../../results/exploratory/{str(i).zfill(4)}_spot/settings.csv")[
            "cfg_model"
        ].item()
        + " | Samples: "
        + str(len(df["top_1"]))
        + " | Failed: "
        + str(df["top_1"].isna().sum().item())
    )
    sns.histplot(df["top_1"], bins=500)
    plt.tight_layout()
    plt.savefig(f"../../results/exploratory/{str(i).zfill(4)}_spot/hisplot.svg")
```
