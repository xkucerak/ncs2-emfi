```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
path = Path("../../results/exploratory/").resolve()
```

```python
for dir in path.iterdir():

    if not int(dir.stem.split("_")[0]) in range(0, 300):
        continue

    results = dir / "results.csv"

    if not results.exists():
        continue

    settings = dir / "settings.csv"

    for type in ["top_1", "top_5"]:
        df_results = pd.read_csv(results).replace(-1, np.nan)
        df_settings = pd.read_csv(settings)
        plt.figure(figsize=(10, 4))
        print()
        plt.title(
            type.upper().replace("_", " ")
            + " | "
            + df_settings["cfg_model"].item().split("/")[-1]
            + " | Image Count: "
            + str(df_settings["cfg_img_count"].item())
            + " | Samples: "
            + str(len(df_results[type]))
            + " | Failed: "
            + str(df_results[type].isna().sum().item())
            + "\n"
            + f"X: {df_settings.pos_x.item()} mm | Y: {df_settings.pos_y.item()} mm | Z: {df_settings.pos_z.item()} mm | U: {df_settings.cs_voltage.item()} V | Delay: {df_settings.cfg_delay.item()} s | Width: {df_settings.cs_width.item()} ns"
        )
        sns.histplot(df_results[type], bins=2**8, binrange=(0, 1))
        plt.tight_layout()
        plt.savefig((dir / f"{type}.svg"))
```
