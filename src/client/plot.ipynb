```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
for i in range(2,5):
    for type in ["top_1","top_5"]:
        df = pd.read_csv(
            f"../../results/exploratory/{str(i).zfill(4)}_spot/results.csv"
        ).replace(-1, np.nan)
        plt.figure(figsize=(10, 4))
        print()
        plt.title(
            type + " | "
            +
            pd.read_csv(f"../../results/exploratory/{str(i).zfill(4)}_spot/settings.csv")[
                "cfg_model"
            ].item()
            + " | Samples: "
            + str(len(df[type]))
            + " | Failed: "
            + str(df[type].isna().sum().item())
        )
        sns.histplot(df[type], bins=2**8)
        plt.tight_layout()
        plt.savefig(f"../../results/exploratory/{str(i).zfill(4)}_spot/{type}.svg")
```
