```python
import socket

SERVER_IP = "10.0.0.2"
PORT = 5050

s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
s.connect((SERVER_IP, PORT))
```

```python
s.sendall(b"run")
s.recv(1024).decode()
```

```python
s.sendall(b"info")
s.recv(1024).decode()
```

```python
s.sendall(b"size,1024")
s.recv(1024).decode()
s.sendall(b"seed,123")
s.recv(1024).decode()
```

```python
s.sendall(b"log,live_injected")
s.recv(1024).decode()
```

```python
s.sendall(b"test")
s.sendall(b"reload")
s.recv(1024).decode()
s.sendall(b"test")
```

```python
s.sendall(b"log,live_reload")
s.recv(1024).decode()
```

```python
s.close()
```
